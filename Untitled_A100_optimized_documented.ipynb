**RMSNorm layer.** Root-mean-square normalization with a learnable scale. The computation runs in FP32 for stability and is cast back to the input dtype.

```python
class RMSNorm(nn.Module):
    def __init__(self, dim, eps=1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def forward(self, x):
        original_dtype = x.dtype

        x_float = x.float()
        rms = torch.sqrt(
            x_float.pow(2).mean(dim=-1, keepdim=True) + self.eps
        )

        normalized = x_float / rms
        normalized = normalized * self.weight.float()

        return normalized.to(original_dtype)
```

**RoPE frequency precomputation.** Builds the complex rotation factors (one per position and frequency pair) used by rotary positional embeddings.

```python
def precompute_freqs_cis(
    head_dim,
    seq_len,
    theta=10000.0,
    device=None
):
    if head_dim % 2 != 0:
        raise ValueError("RoPE requires an even attention head dimension.")

    frequencies = 1.0 / (
        theta ** (
            torch.arange(0, head_dim, 2, device=device).float()
            / head_dim
        )
    )

    positions = torch.arange(seq_len, device=device).float()
    angles = torch.outer(positions, frequencies)

    freqs_complex = torch.polar(
        torch.ones_like(angles),
        angles
    )

    return freqs_complex
```

**Apply RoPE.** Rotates query/key vectors by treating pairs of channels as complex numbers and multiplying by the precomputed frequencies, which injects position information.

```python
def apply_rope(x, freqs_complex, start_pos=0):
    batch_size, n_heads, seq_len, head_dim = x.shape

    if head_dim % 2 != 0:
        raise ValueError("The attention head dimension must be even.")

    end_pos = start_pos + seq_len

    if end_pos > freqs_complex.shape[0]:
        raise ValueError("RoPE frequencies do not cover these positions.")

    x_float = x.float().contiguous()

    x_pairs = x_float.reshape(
        batch_size,
        n_heads,
        seq_len,
        head_dim // 2,
        2
    )

    x_complex = torch.view_as_complex(x_pairs)

    freqs = freqs_complex[start_pos:end_pos]
    freqs = freqs[None, None, :, :]

    rotated_complex = x_complex * freqs

    rotated_real = torch.view_as_real(rotated_complex)
    rotated = rotated_real.reshape(
        batch_size,
        n_heads,
        seq_len,
        head_dim
    )

    return rotated.to(dtype=x.dtype)
```

**`repeat_kv` helper.** Repeats the key/value heads so their count matches the number of query heads, which is what makes grouped-query attention (GQA) work.

```python
def repeat_kv(x, n_rep):
    if n_rep == 1:
        return x

    batch_size, n_kv_heads, seq_len, head_dim = x.shape

    x = x[:, :, None, :, :].expand(
        batch_size,
        n_kv_heads,
        n_rep,
        seq_len,
        head_dim
    )

    return x.reshape(
        batch_size,
        n_kv_heads * n_rep,
        seq_len,
        head_dim
    )
```

**Attention module.** Sliding-window causal attention with GQA: projects Q/K/V, applies RoPE, repeats KV heads, and runs `scaled_dot_product_attention` using a cached causal + window mask (no per-call mask allocation).

```python
class MistralAttention(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        window_size,
        max_seq_len,
        rope_theta=10000.0
    ):
        super().__init__()

        if d_model % n_heads != 0:
            raise ValueError("d_model must be divisible by n_heads.")

        if n_heads % n_kv_heads != 0:
            raise ValueError("n_heads must be divisible by n_kv_heads.")

        self.d_model = d_model
        self.n_heads = n_heads
        self.n_kv_heads = n_kv_heads
        self.head_dim = d_model // n_heads
        self.n_rep = n_heads // n_kv_heads
        self.window_size = window_size
        self.max_seq_len = max_seq_len

        if self.head_dim % 2 != 0:
            raise ValueError("RoPE requires an even head dimension.")

        self.q_proj = nn.Linear(
            d_model, n_heads * self.head_dim, bias=False
        )
        self.k_proj = nn.Linear(
            d_model, n_kv_heads * self.head_dim, bias=False
        )
        self.v_proj = nn.Linear(
            d_model, n_kv_heads * self.head_dim, bias=False
        )
        self.o_proj = nn.Linear(d_model, d_model, bias=False)

        freqs_complex = precompute_freqs_cis(
            head_dim=self.head_dim,
            seq_len=max_seq_len,
            theta=rope_theta
        )

        self.register_buffer(
            "freqs_complex",
            freqs_complex,
            persistent=False
        )

        positions = torch.arange(max_seq_len)
        allowed = ((positions[None, :] <= positions[:, None]) &
                   (positions[None, :] >= positions[:, None] - window_size + 1))
        self.register_buffer("attention_mask", allowed[None, None], persistent=False)

    def forward(self, x):
        batch_size, seq_len, _ = x.shape

        if seq_len > self.max_seq_len:
            raise ValueError(
                f"Sequence length {seq_len} exceeds "
                f"max_seq_len={self.max_seq_len}."
            )

        # Project: (B, T, D) -> query/key/value heads.
        q = self.q_proj(x).view(
            batch_size, seq_len, self.n_heads, self.head_dim
        ).transpose(1, 2)

        k = self.k_proj(x).view(
            batch_size, seq_len, self.n_kv_heads, self.head_dim
        ).transpose(1, 2)

        v = self.v_proj(x).view(
            batch_size, seq_len, self.n_kv_heads, self.head_dim
        ).transpose(1, 2)

        # Apply rotary embeddings to Q and K before KV repetition.
        q = apply_rope(q, self.freqs_complex)
        k = apply_rope(k, self.freqs_complex)

        k = repeat_kv(k, self.n_rep)
        v = repeat_kv(v, self.n_rep)

        # Reuse a cached mask instead of allocating it every forward call.
        attention_mask = self.attention_mask[:, :, :seq_len, :seq_len]

        attention_output = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attention_mask,
            dropout_p=0.0,
            is_causal=False
        )

        attention_output = attention_output.transpose(1, 2).contiguous()
        attention_output = attention_output.view(
            batch_size, seq_len, self.d_model
        )

        return self.o_proj(attention_output)
```

**SwiGLU feed-forward network.** Gated MLP (`down(silu(gate(x)) * up(x))`) used as the body of each expert.

```python
class SwiGLU(nn.Module):
    def __init__(self, d_model, hidden_dim):
        super().__init__()

        self.gate_proj = nn.Linear(d_model, hidden_dim, bias=False)
        self.up_proj = nn.Linear(d_model, hidden_dim, bias=False)
        self.down_proj = nn.Linear(hidden_dim, d_model, bias=False)

    def forward(self, x):
        gate = F.silu(self.gate_proj(x))
        value = self.up_proj(x)

        return self.down_proj(gate * value)
```

**Mixture-of-Experts layer.** A router scores 8 experts per token, the top-2 are chosen and their weights renormalized, each expert processes only its assigned tokens, and outputs are summed in FP32. Also computes the auxiliary load-balancing loss that encourages even expert usage.

```python
class MoE(nn.Module):
    def __init__(
        self,
        d_model,
        hidden_dim,
        num_experts,
        top_k
    ):
        super().__init__()

        if not 1 <= top_k <= num_experts:
            raise ValueError("top_k must be between 1 and num_experts.")

        self.num_experts = num_experts
        self.top_k = top_k

        # Router: assigns a score to each expert for every token.
        self.router = nn.Linear(
            d_model,
            num_experts,
            bias=False
        )

        # Each expert is an independent SwiGLU feed-forward network.
        self.experts = nn.ModuleList([
            SwiGLU(d_model, hidden_dim)
            for _ in range(num_experts)
        ])

    def forward(self, x):
        # x: (B, T, D)
        batch_size, seq_len, d_model = x.shape

        # Flatten batch and sequence dimensions.
        # flat_x: (B*T, D)
        flat_x = x.reshape(-1, d_model)

        # 1. Compute routing probabilities.
        # router_logits: (B*T, num_experts)
        router_logits = self.router(flat_x)

        # Compute softmax in FP32 for numerical stability.
        router_probs = F.softmax(
            router_logits.float(),
            dim=-1
        )

        # 2. Select the top-k experts for every token.
        # top_weights: (B*T, top_k)
        # top_indices: (B*T, top_k)
        top_weights, top_indices = torch.topk(
            router_probs,
            k=self.top_k,
            dim=-1
        )

        # Renormalize selected weights so they sum to 1 per token.
        top_weights = top_weights / top_weights.sum(
            dim=-1,
            keepdim=True
        ).clamp_min(1e-9)

        # 3. Accumulate expert contributions in FP32.
        # This avoids the Float/BFloat16 index_add_ mismatch.
        output = torch.zeros(
            flat_x.shape,
            device=flat_x.device,
            dtype=torch.float32
        )

        # 4. Process the tokens assigned to each expert.
        for expert_idx, expert in enumerate(self.experts):

            # Find each token and its selected top-k slot
            # where this expert was chosen.
            token_indices, slot_indices = torch.where(
                top_indices == expert_idx
            )

            # Empty token selections are supported by Linear/index_add_.
            # Avoid a separate .numel() conditional per expert.
            # Select tokens assigned to this expert.
            expert_input = flat_x[token_indices]

            # Run the expert network.
            expert_output = expert(expert_input)

            # Get the routing weight for each token-expert pair.
            weights = top_weights[
                token_indices,
                slot_indices
            ]

            # Multiply expert output by its routing weight.
            # Convert to FP32 before accumulation.
            weighted_output = (
                expert_output.float()
                * weights.unsqueeze(-1)
            )

            # Multiple experts can contribute to the same token.
            output.index_add_(
                0,
                token_indices,
                weighted_output
            )

        # 5. Compute the auxiliary load-balancing loss.
        #
        # assignment: fraction of selected routing slots assigned
        # to each expert.
        assignment = F.one_hot(
            top_indices,
            num_classes=self.num_experts
        ).float().mean(dim=(0, 1))

        # importance: average router probability for each expert.
        importance = router_probs.mean(dim=0)

        # Encourage routing probability and token assignments
        # to be distributed across experts.
        aux_loss = self.num_experts * torch.sum(
            importance * assignment.detach()
        )

        # 6. Restore the original tensor dimensions.
        output = output.reshape(
            batch_size,
            seq_len,
            d_model
        )

        # Match the input dtype for the following transformer layer.
        output = output.to(x.dtype)

        return output, aux_loss
```

**Transformer block.** One layer: pre-norm attention with a residual connection, then pre-norm MoE with a residual connection. Returns the hidden state and the layer's auxiliary loss.

```python
class MixtralTransformerBlock(nn.Module):
    def __init__(
        self,
        d_model,
        n_heads,
        n_kv_heads,
        hidden_dim,
        num_experts,
        top_k,
        window_size,
        max_seq_len
    ):
        super().__init__()

        self.attention_norm = RMSNorm(d_model)

        self.attention = MistralAttention(
            d_model=d_model,
            n_heads=n_heads,
            n_kv_heads=n_kv_heads,
            window_size=window_size,
            max_seq_len=max_seq_len
        )

        self.moe_norm = RMSNorm(d_model)

        self.moe = MoE(
            d_model=d_model,
            hidden_dim=hidden_dim,
            num_experts=num_experts,
            top_k=top_k
        )

    def forward(self, x):
        x = x + self.attention(self.attention_norm(x))

        moe_output, aux_loss = self.moe(self.moe_norm(x))
        x = x + moe_output

        return x, aux_loss
```

**Full model (`Mixtral8x7B`).** Token embedding → stack of transformer blocks → final RMSNorm → LM head (weights tied to the embeddings). Returns logits and the mean auxiliary loss across layers.

```python
class Mixtral8x7B(nn.Module):
    def __init__(
        self,
        vocab_size,
        d_model=512,
        n_layers=8,
        n_heads=8,
        n_kv_heads=4,
        hidden_dim=1376,
        num_experts=8,
        top_k=2,
        window_size=256,
        max_seq_len=512
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.d_model = d_model
        self.max_seq_len = max_seq_len

        self.token_embedding = nn.Embedding(
            vocab_size, d_model
        )

        self.layers = nn.ModuleList([
            MixtralTransformerBlock(
                d_model=d_model,
                n_heads=n_heads,
                n_kv_heads=n_kv_heads,
                hidden_dim=hidden_dim,
                num_experts=num_experts,
                top_k=top_k,
                window_size=window_size,
                max_seq_len=max_seq_len
            )
            for _ in range(n_layers)
        ])

        self.final_norm = RMSNorm(d_model)

        self.lm_head = nn.Linear(
            d_model, vocab_size, bias=False
        )

        # Tie the output projection to the token embeddings.
        self.lm_head.weight = self.token_embedding.weight

    def forward(self, input_ids):
        if input_ids.ndim != 2:
            raise ValueError("input_ids must have shape (B, T).")

        if input_ids.shape[1] > self.max_seq_len:
            raise ValueError("Input sequence is longer than max_seq_len.")

        x = self.token_embedding(input_ids)

        total_aux_loss = torch.zeros(
            (), device=x.device, dtype=torch.float32
        )

        for layer in self.layers:
            x, aux_loss = layer(x)
            total_aux_loss = total_aux_loss + aux_loss

        x = self.final_norm(x)
        logits = self.lm_head(x)

        mean_aux_loss = total_aux_loss / len(self.layers)

        return logits, mean_aux_loss
```

**Token cache preparation.** One-time preprocessing: streams FineWeb-Edu documents, tokenizes them, packs tokens (separated by EOS) into fixed-length records of `SEQ_LEN + 1`, and writes them to a binary file on local disk with a metadata file. It reuses a valid existing cache. Also defines `CachedPackedTextDataset`, which reads the cache through a memory map and returns `(inputs, targets)` shifted by one token.

```python
import numpy as np
from torch.utils.data import Dataset

# One-time preprocessing: write fixed-size token sequences to local SSD.
# Repeated epochs will not repeat network streaming or tokenization.
CACHE_PATH = OUTPUT_DIR / f"fineweb_edu_{MAX_DOCUMENTS}_seq{SEQ_LEN}.bin"
CACHE_META_PATH = OUTPUT_DIR / f"fineweb_edu_{MAX_DOCUMENTS}_seq{SEQ_LEN}.json"


def prepare_token_cache():
    if CACHE_PATH.exists() and CACHE_META_PATH.exists():
        metadata = json.loads(CACHE_META_PATH.read_text())
        expected_bytes = metadata["num_samples"] * (SEQ_LEN + 1) * np.dtype("uint32").itemsize
        if CACHE_PATH.stat().st_size == expected_bytes and metadata["num_samples"] > 0:
            print(f"Using existing local cache: {metadata['num_samples']:,} samples")
            return
        print("Invalid/incomplete cache found; rebuilding it.")

    source = load_dataset(DATASET_NAME, DATASET_CONFIG, split="train", streaming=True)
    eos = tokenizer.eos_token_id
    buffer = []
    samples = 0
    temp_path = CACHE_PATH.with_suffix(".partial")
    with open(temp_path, "wb") as fout:
        for doc_idx, row in enumerate(tqdm(source, total=MAX_DOCUMENTS, desc="Caching FineWeb-Edu")):
            if doc_idx >= MAX_DOCUMENTS:
                break
            text = row.get("text", "")
            if not isinstance(text, str) or not text.strip():
                continue
            ids = tokenizer(text, add_special_tokens=False, truncation=False, verbose=False)["input_ids"]
            if not ids:
                continue
            buffer.extend(ids)
            if eos is not None:
                buffer.append(eos)
            offset = 0
            while len(buffer) - offset >= SEQ_LEN + 1:
                np.asarray(buffer[offset:offset + SEQ_LEN + 1], dtype=np.uint32).tofile(fout)
                samples += 1
                offset += SEQ_LEN
            if offset:
                del buffer[:offset]
    if not samples:
        raise RuntimeError("No training samples generated from the dataset")
    os.replace(temp_path, CACHE_PATH)
    CACHE_META_PATH.write_text(json.dumps({"num_samples": samples, "seq_len": SEQ_LEN}))
    print(f"Saved {samples:,} samples in {CACHE_PATH}")


class CachedPackedTextDataset(Dataset):
    def __init__(self, path, seq_len):
        self.path = Path(path)
        self.seq_len = seq_len
        record_bytes = (seq_len + 1) * np.dtype("uint32").itemsize
        assert self.path.stat().st_size % record_bytes == 0, "Invalid cache file"
        self.num_samples = self.path.stat().st_size // record_bytes
        self._tokens = None  # Lazily open in each dataloader worker

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        if self._tokens is None:
            self._tokens = np.memmap(self.path, mode="r", dtype=np.uint32,
                                     shape=(self.num_samples, self.seq_len + 1))
        ids = torch.from_numpy(np.array(self._tokens[idx], dtype=np.int64))
        return ids[:-1], ids[1:]


import json
prepare_token_cache()
```

**Learning-rate schedule.** Linear warmup for 200 steps, then cosine decay toward 10% of the base learning rate over a 100,000-step horizon, wired in through `LambdaLR`.

```python
def lr_lambda(step):
    if step < WARMUP_STEPS:
        return max(1, step + 1) / max(1, WARMUP_STEPS)

    # The runtime limit controls training duration; this schedule
    # defines a long, gradual decay over up to 100,000 optimizer steps.
    total_steps = 100_000
    progress = min(
        1.0,
        (step - WARMUP_STEPS) /
        max(1, total_steps - WARMUP_STEPS)
    )

    return 0.1 + 0.9 * 0.5 * (1.0 + math.cos(math.pi * progress))


scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer,
    lr_lambda=lr_lambda
)
```

**Batch helper.** `get_next_batch()` returns the next batch and restarts the iterator when an epoch ends.

```python
def get_next_batch():
    global train_iterator

    try:
        return next(train_iterator)

    except StopIteration:
        train_iterator = iter(train_loader)

        try:
            return next(train_iterator)

        except StopIteration as exc:
            raise RuntimeError(
                "The dataset produced no batches. Check the dataset "
                "configuration and document limit."
            ) from exc
```

**Checkpoint saving.** `save_checkpoint()` writes model, optimizer, scheduler, step, loss history, and config to disk (via a temp file and atomic rename so a crash can't leave a corrupt checkpoint).

```python
def save_checkpoint(
    model,
    optimizer,
    scheduler,
    step,
    loss_history,
    checkpoint_dir=CHECKPOINT_DIR
):
    checkpoint_dir = Path(checkpoint_dir)
    checkpoint_dir.mkdir(parents=True, exist_ok=True)

    checkpoint_path = checkpoint_dir / f"checkpoint_step_{step}.pt"
    temporary_path = checkpoint_dir / f"checkpoint_step_{step}.tmp"

    checkpoint = {
        "step": step,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": scheduler.state_dict(),
        "loss_history": loss_history,
        "config": {
            "vocab_size": VOCAB_SIZE,
            "d_model": D_MODEL,
            "n_layers": N_LAYERS,
            "n_heads": N_HEADS,
            "n_kv_heads": N_KV_HEADS,
            "hidden_dim": HIDDEN_DIM,
            "num_experts": NUM_EXPERTS,
            "top_k": TOP_K,
            "window_size": WINDOW_SIZE,
            "seq_len": SEQ_LEN
        }
    }

    torch.save(checkpoint, temporary_path)
    os.replace(temporary_path, checkpoint_path)

    print(f"\nCheckpoint saved: {checkpoint_path}")

    return checkpoint_path
```

**Quick inference test.** `run_inference_test()` greedily generates 30 tokens from a fixed prompt using BF16 autocast, prints the sample, and restores the model's previous train/eval mode. It's used for a progress check after each epoch.

```python
INFERENCE_PROMPT = "Artificial intelligence is transforming the world because"

@torch.inference_mode()
def run_inference_test(model, tokenizer, prompt=INFERENCE_PROMPT):
    # This custom model does not implement Hugging Face model.generate().
    was_training = model.training
    model.eval()
    ids = tokenizer(prompt, return_tensors="pt")["input_ids"].to(DEVICE)
    try:
        for _ in range(30):
            with torch.autocast("cuda", dtype=torch.bfloat16):
                logits, _ = model(ids[:, -model.max_seq_len:])
            next_id = logits[:, -1].argmax(dim=-1, keepdim=True)
            ids = torch.cat([ids, next_id], dim=1)
            if next_id.item() == tokenizer.eos_token_id:
                break
        print("\nSample:", tokenizer.decode(ids[0], skip_special_tokens=True))
    finally:
        model.train(was_training)
```

**Single training-step smoke test.** Runs one forward pass, loss computation (cross-entropy + weighted aux loss), backward pass, and optimizer step. It asserts that the loss and gradients are finite before the long run starts.

```python
model.train()
optimizer.zero_grad(set_to_none=True)

batch = next(iter(train_loader))
input_ids, target_ids = batch

input_ids = input_ids.to(DEVICE)
target_ids = target_ids.to(DEVICE)

with torch.autocast("cuda", dtype=torch.bfloat16):
    logits, aux_loss = model(input_ids)

    main_loss = F.cross_entropy(
        logits.reshape(-1, logits.size(-1)).float(),
        target_ids.reshape(-1)
    )

    loss = main_loss + AUX_LOSS_COEF * aux_loss

assert torch.isfinite(loss), "Loss is not finite!"

loss.backward()

# Confirm that at least one parameter received a gradient.
has_gradient = any(
    p.grad is not None and torch.isfinite(p.grad).all()
    for p in model.parameters()
    if p.requires_grad
)

assert has_gradient, "No finite gradients found!"

optimizer.step()

print("Forward pass: OK")
print("Backward pass: OK")
print("Optimizer step: OK")
print("Main loss:", main_loss.item())
print("Auxiliary loss:", aux_loss.item())
print("Total loss:", loss.item())
```

**Training loop (`train_model`).** Runs until the time limit: BF16 autocast forward/backward with gradient accumulation, gradient clipping, optimizer and scheduler steps, GPU-side loss accumulation to avoid per-batch syncs, periodic logging, timed checkpoints, handling of a partial accumulation at the end of an epoch, an inference sample after each epoch, and a final checkpoint. Returns the loss history.

```python
import time


def train_model(model, optimizer, scheduler, max_minutes, grad_accum_steps,
                checkpoint_interval_seconds):
    model.train()
    start_time = time.time()
    deadline = start_time + max_minutes * 60
    last_checkpoint_time = start_time
    loss_history = []
    optimizer_steps = 0
    total_microbatches = 0
    epoch = 0
    optimizer.zero_grad(set_to_none=True)

    while time.time() < deadline:
        epoch += 1
        pending = 0
        epoch_batches = 0
        epoch_loss = 0.0
        log_loss_tensor = torch.zeros((), device=DEVICE)
        bar = tqdm(train_loader, desc=f"Epoch {epoch}", mininterval=2.0)
        for input_ids, target_ids in bar:
            if time.time() >= deadline:
                break
            input_ids = input_ids.to(DEVICE, non_blocking=True)
            target_ids = target_ids.to(DEVICE, non_blocking=True)
            with torch.autocast("cuda", dtype=torch.bfloat16):
                logits, aux_loss = model(input_ids)
                main_loss = F.cross_entropy(logits.flatten(0, 1).float(),
                                            target_ids.flatten())
                loss = main_loss + AUX_LOSS_COEF * aux_loss
            (loss / grad_accum_steps).backward()
            # Accumulate on GPU; no .item() GPU-to-CPU synchronization per batch.
            log_loss_tensor += loss.detach()
            pending += 1
            epoch_batches += 1
            total_microbatches += 1

            if pending == grad_accum_steps:
                torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP_NORM)
                optimizer.step()
                scheduler.step()
                optimizer.zero_grad(set_to_none=True)
                optimizer_steps += 1
                pending = 0

                if optimizer_steps % LOG_EVERY_UPDATES == 0:
                    avg_loss = (log_loss_tensor / (LOG_EVERY_UPDATES * grad_accum_steps)).item()
                    log_loss_tensor.zero_()
                    loss_history.append({"step": optimizer_steps,
                                         "lm_loss": avg_loss,
                                         "total_loss": avg_loss})
                    bar.set_postfix(loss=f"{avg_loss:.4f}", steps=optimizer_steps)

                now = time.time()
                if now - last_checkpoint_time >= checkpoint_interval_seconds:
                    save_checkpoint(model=model, optimizer=optimizer, scheduler=scheduler,
                                    step=optimizer_steps, loss_history=loss_history)
                    last_checkpoint_time = now

        bar.close()
        if epoch_batches == 0:
            break
        # Correct loss scaling for any partial accumulation at epoch/end of time.
        if pending:
            for p in model.parameters():
                if p.grad is not None:
                    p.grad.mul_(grad_accum_steps / pending)
            torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP_NORM)
            optimizer.step()
            scheduler.step()
            optimizer.zero_grad(set_to_none=True)
            optimizer_steps += 1

        print(f"Epoch {epoch}: {epoch_batches} microbatches; {optimizer_steps} total updates")
        if time.time() >= deadline:
            break
        run_inference_test(model, tokenizer)

    save_checkpoint(model=model, optimizer=optimizer, scheduler=scheduler,
                    step=optimizer_steps, loss_history=loss_history)
    print(f"Training finished. Updates={optimizer_steps}, microbatches={total_microbatches}, "
          f"minutes={(time.time()-start_time)/60:.2f}")
    return loss_history
```

**Text generation function.** `generate()` performs autoregressive sampling with temperature and top-k filtering (or greedy decoding when temperature is 0), crops the context to `max_seq_len`, and stops at EOS.

```python
@torch.inference_mode()
def generate(
    model,
    tokenizer,
    prompt,
    max_new_tokens=100,
    temperature=0.8,
    top_k=50
):
    if temperature < 0:
        raise ValueError("temperature must be non-negative.")

    if top_k < 1:
        raise ValueError("top_k must be at least 1.")

    model.eval()

    input_ids = tokenizer.encode(
        prompt,
        return_tensors="pt"
    ).to(DEVICE)

    if input_ids.shape[1] == 0:
        raise ValueError("The prompt must contain at least one token.")

    for _ in range(max_new_tokens):
        # Keep only the most recent context tokens.
        context = input_ids[:, -model.max_seq_len:]

        with torch.autocast(
            device_type="cuda",
            dtype=torch.bfloat16
        ):
            logits, _ = model(context)

        next_token_logits = logits[:, -1, :].float()

        if temperature == 0:
            next_token = torch.argmax(
                next_token_logits,
                dim=-1,
                keepdim=True
            )
        else:
            next_token_logits = next_token_logits / temperature

            k = min(top_k, next_token_logits.shape[-1])
            top_values, _ = torch.topk(next_token_logits, k=k, dim=-1)

            cutoff = top_values[:, -1].unsqueeze(-1)

            next_token_logits = next_token_logits.masked_fill(
                next_token_logits < cutoff,
                float("-inf")
            )

            probabilities = F.softmax(next_token_logits, dim=-1)

            next_token = torch.multinomial(
                probabilities,
                num_samples=1
            )

        input_ids = torch.cat(
            [input_ids, next_token],
            dim=1
        )

        if (
            tokenizer.eos_token_id is not None
            and next_token.item() == tokenizer.eos_token_id
        ):
            break

    return tokenizer.decode(
        input_ids[0],
        skip_special_tokens=True
    )
```

**Checkpoint loading.** `load_training_checkpoint()` restores model weights, and optionally optimizer and scheduler state, from a saved checkpoint so training can be resumed.

```python
def load_training_checkpoint(
    checkpoint_path,
    model,
    optimizer=None,
    scheduler=None
):
    checkpoint = torch.load(
        checkpoint_path,
        map_location="cpu",
        weights_only=False
    )

    model.load_state_dict(checkpoint["model_state_dict"])

    if optimizer is not None:
        optimizer.load_state_dict(
            checkpoint["optimizer_state_dict"]
        )

    if scheduler is not None:
        scheduler.load_state_dict(
            checkpoint["scheduler_state_dict"]
        )

    print("Restored step:", checkpoint["step"])

    return checkpoint
```
